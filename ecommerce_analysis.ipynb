# E-Commerce Data Analysis with Pandas

This notebook samples 10% of the UCI Online Retail transactions, cleans missing and invalid values, calculates revenue, and compares the top 10 countries.

If the workbook is not present, run `python download_data.py` from the project directory first.

## 1. Load a reproducible 10% sample

```python
import matplotlib.pyplot as plt
from analyze_retail import clean_transactions, load_sample, revenue_by_country, save_results

sample = load_sample()
print(f'Sampled {len(sample):,} of the source rows ({len(sample) / 541_909:.1%}).')
sample.head()
```

## 2. Clean transaction records

The cleaning function drops rows missing required fields, converts dates and numeric values, filters canceled invoices and non-positive quantities (returns), and excludes zero or negative prices (free items).

```python
cleaned = clean_transactions(sample)
cleaned['Revenue'] = cleaned['Quantity'] * cleaned['UnitPrice']
print(f'Rows after cleaning: {len(cleaned):,} (removed {len(sample) - len(cleaned):,})')
cleaned.dtypes
```

```python
cleaned.head()
```

## 3. Top 10 countries by total revenue

```python
top_countries = revenue_by_country(cleaned)
top_countries.style.format({'Revenue': '£{:,.2f}'})
```

```python
chart_data = top_countries.sort_values('Revenue')
fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(chart_data['Country'], chart_data['Revenue'], color='#2f6fed')
ax.set_title('Top 10 Countries by Revenue (10% Sample)')
ax.set_xlabel('Revenue (£)')
ax.set_ylabel('Country')
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda value, _: f'£{value:,.0f}'))
ax.grid(axis='x', alpha=0.2)
fig.tight_layout()
plt.show()
```

## 4. Save reusable outputs

```python
save_results(cleaned, top_countries)
print('Saved cleaned sample, revenue table, and chart under outputs/.')
```
